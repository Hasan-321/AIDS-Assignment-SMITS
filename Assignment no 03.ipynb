### Question 5: Number Analyzer

### Question 9: Student Marks Entry System

### Question 10: Restaurant Ordering System 
You are developing a simple restaurant ordering system.

### Question 11: Student Performance Management System 
A teacher wants to create a system for analyzing the performance of multiple students. 
First ask: 
Enter number of students: 
For each student, ask: 
• Student name  
• Marks in Python  
• Marks in Mathematics  
• Marks in English  
Calculate: 
Total Marks 
Percentage 
Result Rules 
A student is considered Pass only if: 
• Overall percentage is at least 50%  
• AND marks in every subject are at least 40.  
Otherwise, the student is Fail. 
Grade Rules 
For students who pass: 
90+     → A+ 
80-89   → A 
70-79   → B 
60-69   → C 
50-59   → D 
While processing students, calculate: 
• Number of passed students  
• Number of failed students  
• Highest percentage  
• Lowest percentage  
• Class total marks  
• Class average percentage  
Final Display: 
===== CLASS REPORT ===== 
Total Students: 10 
Passed: 7 
Failed: 3 
Highest Percentage: 94% 
Lowest Percentage: 42% 
Class Average: 76.5% 
Important requirement 
Use a loop to process all students. 
Do not manually write the same code for every student.


```python
number_of_Student = int(input("Enter a number of student: "))
pass_student = 0
fail_student = 0
highest_percentage = 0
lowest_percentage = 100
class_total_marks = 0
class_total_percentage = 0
for i in range(number_of_Student):
    print(f"Enter a detail of student {i + 1}")

    name_Student = input("Enter your name:")
    marks_python = int(input("Enter python marks: "))
    marks_english = int(input("Enter english marks: "))
    marks_math = int(input("Enter math marks: "))
    total_marks = marks_python + marks_english + marks_math
    percentage = (total_marks/300)*100

    if (percentage >=50 and marks_python >=40 and marks_english >= 40 and marks_math >= 40):
        result = "Pass"
        pass_student += 1
        if (percentage >= 90):
            Grade = "A+"
        elif (percentage >=80 and percentage <= 89):
            Grade = "A"
        elif (percentage >=70 and percentage <= 79):
            Grade = "B"
        elif (percentage >=60 and percentage <= 69):
            Grade = "C"
        else:
            Grade = "D"
    else:
        result = "Fail"
        Grade = "N/A"
        fail_student += 1
    print("======Student Report======")
    print(f"Name: {name_Student}")
    print(f"Total marks: {total_marks} / 300")
    print(f"Percentage: {percentage}%")
    print(f"Result: {result}")
    print(f"Grade: {Grade}")

    class_total_marks += total_marks
    class_total_percentage += percentage

    if (i == 0):
        highest_percentage = percentage
        lowest_percentage = percentage
    else:
        if (percentage > highest_percentage):
            highest_percentage = percentage
        elif (percentage< lowest_percentage):  
            lowest_percentage = percentage
if (number_of_Student > 0):
    class_avg = class_total_percentage/number_of_Student
else: 
    class_avg = 0

print("===== CLASS REPORT =====")
print(f"Total Students: {number_of_Student}")
print(f"Passed: {pass_student}")
print(f"Failed: {fail_student}")
print(f"Highest Percentage: {highest_percentage} %")
print(f"Lowest Percentage: {lowest_percentage} %")
print(f"Class Total Marks: {class_total_marks}")
print(f"Class Average:{class_avg}%")
```
